```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# 1. Upload & Import Data
files.upload()
df = pd.read_excel("yahya_agus.xlsx") # <--- Menggunakan nama file baru
df.columns = df.columns.str.strip()

# 2. Tabel Frekuensi
frq = df["Grade"].value_counts()
print("=== TABEL FREKUENSI ===\n", frq)

# 3. Visualisasi Data
plt.style.use('seaborn-v0_8-whitegrid')

# Line Chart (Warna Hijau Teal)
plt.figure(figsize=(6, 3))
plt.plot(frq.index, frq.values, marker='o', color='#006d77', lw=2)
plt.title('Distribusi Grade (Line)')
plt.show()

# Bar Chart (Warna Ungu Soft)
plt.figure(figsize=(6, 3))
plt.bar(frq.index, frq.values, color='#8338ec')
plt.title('Distribusi Grade (Bar)')
plt.show()

# Pie Chart (Skema Warna Custom Modern)
custom_colors = ['#006d77', '#8338ec', '#3a86ff', '#ff006e', '#fb5607', '#ffbe0b']
plt.figure(figsize=(4, 4))
plt.pie(frq.values, labels=frq.index, autopct='%1.0f%%', colors=custom_colors[:len(frq)])
plt.title('Proporsi Grade (Pie)')
plt.show()

# 4. Statistika Deskriptif
col = [c for c in df.select_dtypes('number') if 'NIM' not in c.upper() and 'ID' not in c.upper()][-1]
dt = pd.to_numeric(df[col], errors='coerce')

stats = dt.describe()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode()[0]
stats['Variance'] = dt.var()
stats['Skewness'] = dt.skew()

print(f"\n=== STATISTIKA DESKRIPTIF ({col}) ===")
print(stats)
```
